```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
import plotly.express as px
warnings.filterwarnings('ignore')
```

# **EXPLORACIÓN DE ESTRUCTURA**

```python
df = pd.read_csv(r'C:\Users\pedro\OneDrive\Escritorio\Proyectos\Mapa Secundarios Primera Sección Electoral\Dataset\dataset.csv')
df
```

```python
df['nivel'].unique()
```

```python
municipios_primera = ['Campana', 'Escobar', 'General Las Heras', 'General Rodríguez',
    'General San Martín', 'Hurlingham', 'Ituzaingó', 'José C. Paz',
    'Luján', 'Malvinas Argentinas', 'Marcos Paz', 'Mercedes',
    'Merlo', 'Moreno', 'Morón', 'Navarro', 'Pilar', 'San Fernando',
    'San Isidro', 'San Miguel', 'Suipacha', 'Tigre', 'Tres de Febrero',
    'Vicente López'
]

condicion_nivel = (df['nivel'] == 'Nivel Secundario') | (df['nivel'] == 'Ciclo Medio')

condicion_sector = (df['sector'] == 'Estatal')

condicion_municipio = (df['municipio_nombre'].isin(municipios_primera))


df_eda = df[condicion_nivel & condicion_sector & condicion_municipio]
```

# **GRÁFICOS**

```python
sns.set_theme(style="whitegrid", font="sans-serif")
plt.rcParams['axes.titlesize'] = 16
plt.rcParams['axes.titleweight'] = 'bold'
plt.rcParams['axes.labelsize'] = 12
custom_palette = sns.color_palette("viridis", as_cmap=False)
```

**Ranking de Escuelas por Municipio**

```python
plt.figure(figsize=(10, 8))
ax = sns.countplot(y='municipio_nombre', data=df_eda,
                   order=df_eda['municipio_nombre'].value_counts().index,
                   palette="viridis", hue='municipio_nombre', legend=False)

plt.title('Cantidad de Escuelas por Municipio', pad=20)
plt.xlabel('Número de establecimientos')
plt.ylabel('Municipio')
sns.despine(left=True, bottom=True) # Elimina bordes para un look más limpio
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

**Distribución de Desfavorabilidad**

```python
plt.figure(figsize=(14, 7))

sns.boxplot(x='municipio_nombre', y='desfavorabilidad', data=df_eda, color="skyblue", boxprops=dict(alpha=.5))
sns.stripplot(x='municipio_nombre', y='desfavorabilidad', data=df_eda, color="darkblue", size=4, alpha=0.3)

plt.title('Distribución de Índices de Desfavorabilidad', pad=20)
plt.xticks(rotation=45, ha='right')
plt.xlabel('')
plt.ylabel('Índice de Desfavorabilidad')
plt.show()
```

**Matrícula: Varones vs Mujeres**

```python
plt.figure(figsize=(10, 5))
sns.histplot(df_eda['matricula'].dropna(), bins=40, kde=True,
             color="#2a9d8f", edgecolor="white", alpha=0.7, linewidth=0.5)

plt.title('Distribución del Tamaño de la Matrícula', pad=20)
plt.xlabel('Cantidad de Alumnos')
plt.ylabel('Frecuencia')
sns.despine() # Estética minimalista
plt.show()
```
